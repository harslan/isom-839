## The Sawyer Furniture Co.

The company makes **tables** and **chairs**:

| | Profit | Carpentry | Finishing |
|---|---|---|---|
| Table | $120 | 4 hrs | 2 hrs |
| Chair | $80 | 3 hrs | 1 hr |
| **Available this week** | | **240 hrs** | **100 hrs** |

**Before you run anything:** write your intuitive production plan here 👇 (double-click this cell to edit)

- My guess: ____ tables and ____ chairs → profit $ ______

## The model, in the three-ingredient grammar

**Decision variables:** $T$ = tables to build, $C$ = chairs to build

**Objective:** maximize $120T + 80C$

**Constraints:**
- Carpentry: $4T + 3C \le 240$
- Finishing: $2T + 1C \le 100$
- $T, C \ge 0$

```python
m = gp.Model('sawyer_furniture')

# 1 — Decision variables
T = m.addVar(name='tables')
C = m.addVar(name='chairs')

# 2 — Objective
m.setObjective(120*T + 80*C, GRB.MAXIMIZE)

# 3 — Constraints
carpentry = m.addConstr(4*T + 3*C <= 240, 'carpentry_hours')
finishing = m.addConstr(2*T + 1*C <= 100, 'finishing_hours')

m.optimize()

print()
print(f'Build {T.X:.1f} tables and {C.X:.1f} chairs')
print(f'Maximum weekly profit: ${m.ObjVal:,.0f}')
```

**How did your guess do?** Most first guesses land 10–20% below the optimum — and that gap, at company scale, is why this field has documented **$431 billion+** in impact.

Notice the solver didn't just find *a good* answer. It found the **provably best** answer — a *certificate of optimality*. That guarantee is what companies pay for.

### A preview of Session 3: what is one more hour worth?

The solver also computes **shadow prices** — the dollar value of one more unit of each scarce resource:

```python
print(f'One extra carpentry hour is worth: ${carpentry.Pi:.2f}')
print(f'One extra finishing hour is worth: ${finishing.Pi:.2f}')
print()
print('If overtime costs $25/hour, in which department should you buy it?')
```
