### Mathematical Formulation
**Decision Variables:**
* $x_1$: Number of batches of bread to produce
* $x_2$: Number of batches of cake to produce

**Objective Function:**
* Maximize Profit: $Z = 50x_1 + 60x_2$

**Constraints:**
* Demand for bread: $x_1 \le 90$
* Demand for cake: $x_2 \le 60$
* Oven time (minutes): $60x_1 + 45x_2 \le 4650$
* Labor time (minutes): $30x_1 + 60x_2 \le 4200$
* Non-negativity: $x_1, x_2 \ge 0$

```python
import gurobipy as gp
from gurobipy import GRB

# Create a new model
model = gp.Model("bakery_production")

# Create decision variables
# We assume the number of batches should be an integer.
# If partial batches are allowed, you can change vtype to GRB.CONTINUOUS
bread = model.addVar(vtype=GRB.INTEGER, name="bread", lb=0, ub=90)
cake = model.addVar(vtype=GRB.INTEGER, name="cake", lb=0, ub=60)

# Set the objective function to maximize profit
model.setObjective(50 * bread + 60 * cake, GRB.MAXIMIZE)

# Add constraint for Oven time
model.addConstr(60 * bread + 45 * cake <= 4650, "oven_time")

# Add constraint for Labor time
model.addConstr(30 * bread + 60 * cake <= 4200, "labor_time")

# Optimize the model
model.optimize()

# Print the results
if model.status == GRB.OPTIMAL:
    print('\n--- Optimal Production Plan ---')
    for v in model.getVars():
        print(f'{v.varName}: {v.x} batches')
    print(f'Maximum Profit: ${model.objVal}')
else:
    print('Optimal solution was not found.')
```
